```python
# data serialization
import json

data = {
    'name':'Sapthasree',
    'age':21 
}

# converting to json
json_str = json.dumps(data)
print(json_str)
print(type(json_str))

# convert it back to dictionary
parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
# csv
import csv

with open('example.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name', 'age'])
    writer.writerow(['Sapthasree', 21])

with open('example.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
# datetime
from datetime import datetime, timedelta

# today's date
now = datetime.now()
print(now)

# yesterday's date
ystrdy = now - timedelta(days=1)
print(ystrdy)
```

```python
# time
import time

print(time.time())
time.sleep(2)
print(time.time())
```

```python
# regular expression - used to match a particular expression in a sentence
import re

# for finding digits 
pattern = r'\d+'
text = 'There are 145 apples'
match = re.search(pattern, text)
print(match.group())
```
