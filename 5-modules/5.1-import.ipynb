Modules and packages helps organize and reuse code.

```python
import math
math.sqrt(5)
```

```python
from math import sqrt, pi
print(sqrt(25))
print(pi)
```

```python
import numpy as np
np.array([1,2,3,4,5])
```

```python
from math import *
sqrt(16)
```

```python
# importing from our own package
from package.maths import addition
addition(5, 10)
```

```python
from package import maths

maths.addition(3,4)
```
