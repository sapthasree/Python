Tuples

- ordered collection of items
- immutable
- similar to lists

```python
# create a tuple
empty_tuple = ()
print(empty_tuple, type(empty_tuple))
```

```python
nums = tuple([1,2,3,4,5])
print(nums, type(nums))
```

```python
# accessing tuple elements
print(nums[3])
print(nums[3:])
print(nums[:3])
print(nums[-3])
print(nums[-3:])
print(nums[:-3])
```

```python
# tuple operations

nums2 = (1, 4.0, True, 32)
print(nums)
print(nums2)
```

```python
# concatenation
concat_tuple = nums + nums2
print(concat_tuple, type(concat_tuple))
```

```python
# replication
nums2 * 3
```

```python
# immutable nature

nums[1] = 3 #TypeError
```

```python
# tuple methods

print(nums.count(2)) # count how many times 2 appeared in the tuple
print(nums.index(5)) # generate index position of element 5
```

packing and unpacking tuple

```python
packed_tuple = 1, "Hello", 3.14
print(packed_tuple)
```

```python
# unpacking a tuple

a, b, c = packed_tuple

print(a)
print(b)
print(c)
```

```python
# unpacking with *

numbers = (1,2,3,4,5,6)
first, *middle, last = numbers

print(first)
print(middle)
print(last)
```

```python
# nested tuple

nested_tuple = ((1,2,3), ("a", "b", "c"), (True, False))
```

```python
# accessing elements in a tuple

print(nested_tuple[2])
print(nested_tuple[2][1])
print(nested_tuple[1][2])
```

```python
# iterating over nested tuple 

for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item, end=" ")
    print()
```
