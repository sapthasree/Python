##### Real world example using lists in python

Example 1. Manage a to-do list

```python
to_do_list = ["buy groceries", "clean house", "pay bills"]
print("To do list initially")
for task in to_do_list:
    print(f"-{task}")

# adding to task
to_do_list.append("schedule meeting")
to_do_list.append("go for gym")
print()
print("Updated to do list")
for task in to_do_list:
    print(f"-{task}")

# removing a completed task
to_do_list.remove("clean house")
print("Updated to do list")
print()
for task in to_do_list:
    print(f"-{task}")

# checking if the task is in the list
if "pay bills" in to_do_list:
    print("Don't forget to pay the utility bills")

# to do list remaining
print()
print("To do list remaining")
for task in to_do_list:
    print(f"-{task}")
```

Example 2. Organizing Student Grades

```python
# organizing student grades
grades = [85, 92, 78, 90, 88]

# adding a new grade
grades.append(95)

# calculating avg grade
avg_grade = sum(grades) / len(grades)
print(f"Average-Grade: {avg_grade:.2f}")

# finding the highest and lowest grades
high = max(grades)
low = min(grades)
print(f"Highest Grade: {high}")
print(f"Lowest Grade: {low}")
```

Example 3. Managing an Inventory

- use a list to manage inventory items in a store

```python
inventory = ["apples", "banana", "orange", "grapes"]
print("Initial inventory")
print("inventory list: ")
for item in inventory:
    print(f"- {item}")


# adding a new item
inventory.append("strawberry")
print()
print("Updated inventory")
print("inventory list: ")
for item in inventory:
    print(f"- {item}")

# removing an item which is out of stock
inventory.remove("banana")
print()
print("Updated inventory")
print("inventory list: ")
for item in inventory:
    print(f"- {item}")

# check if an item is out of stock
item = "oranges"
if item in inventory:
    print(f"{item} are in stock")
else:
    print(f"{item} are out of stock")

# printing the inventory
print()
print("Final Inventory ")
for item in inventory:
    print(f"- {item}")
```

Example 4. Collecting user feedback

- use a list to collect and analyze user feedback

```python
feedback = ["Great service!", "Very Satisfied", "Could be better", "Excellent experience"]
print("Initial feedback")
for comment in feedback:
    print(f"- {comment}")

# adding a new feedback
feedback.append("Not happy with the service")
print()
print("updated feedback")
for comment in feedback:
    print(f"- {comment}")

# counting specific feedback
postive_feedback = sum(1 for comment in feedback if "great" in comment.lower() or "excellent" in comment.lower() or "satisfied" in comment.lower())
print()
print(f"Positive Feedback count: {postive_feedback}")

# printing user feedback
print()
print("Final feedback")
for comment in feedback:
    print(f"- {comment}")
```
