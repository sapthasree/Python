Dictionaries

- unordered collection of items
- store data in the form of key-value pairs
- keys must be unique and immutable
- values can be of any type

```python
# creating dict

empty_dict = {}
print(empty_dict, type(empty_dict))
```

```python
empty_dict = dict()
print(empty_dict, type(empty_dict))
```

```python
student = {"name":"saptha", "age":21, "student":True}
print(student, type(student))
```

```python
# error case
student = {"name":"saptha", "age":21, "student":True, "student":False}
print(student)
```

```python
# accessing dict
print(student['name'])
print(student['age'])

# accessing using get method
print(student.get('student'))
print(student.get('grade', 'Not available'))
```

```python
# modifying dict elements

student['age'] = 22 
print(student)
student['grade'] = 'A'
print(student)
del student['student']
print(student)
```

```python
# dict methods

keys = student.keys()
print(keys)

vals = student.values()
print(vals)

items = student.items()
print(items)
```

```python
student_copy = student
print(student)
print(student_copy)
```

```python
student['name'] = 'Sree'

print(student)
print(student_copy)
```

```python
# shallow copy

student_copy1 = student.copy()
print(student)
print(student_copy1)
```

```python
student['name'] = 'Sapthasree'
print(student)
print(student_copy1)
```

```python
# iterating over dictionaries
# use loops to iterate through dict, keys, values or items
```

```python
# iterating over keys
for keys in student.keys():
    print(keys)
```

```python
# iterating over values
for val in student.values():
    print(val)
```

```python
# iterating over key-value pairs
for key, val in student.items():
    print(f"{key}:{val}")
```

```python
# nested dictionary
students = {
    1:{"name":"saptha", "age":22},
    2:{"name":"sree", "age":20},
    3:{"name":"sapthasree", "age":21}
}
print(students)
```

```python
# accessing nested dict
print(students[1]["name"])
print(students[3]["age"])
```

```python
# iterating over nested dict
for id, info in students.items():
    print(f'{id}:{info}')
    for key, val in info.items():
        print(f'{key}:{val}')
    print()
```

```python
# dictionary comprehension
squares = {x: x**2 for x in range(5)}
print(squares)
```

```python
# dictionary comprehension with condition
even_squares = {x: x**2 for x in range(10) if x % 2 == 0}
print(even_squares)
```

Practical Examples

```python
# use a dict to count the number of occurences of elements in a list

nums = [1,2,1,1,1,2,3,2,4,3]
counts = {}

for num in nums:
    if num in counts:
        counts[num] += 1
    else:
        counts[num] = 1

print(counts)
```

```python
# merge two dicts
dict1 = {1:"hello", 2:3.14, 3:"hey"}
dict2 = {5:"heyyy", 4:2000, 7:[2,5,6]}

merged_dict = {**dict1, **dict2}
print(merged_dict)
```
