##### File operation - Read and write files

Python provides built in functions and methods to read from and write to files, both text and binary. 

#### Text Files

```python
# read a whole file

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
# read a file line by line

with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip()) # strip() used to remove extra line
```

```python
# writing a file(Overwriting)

with open('example.txt', 'w') as file:
    file.write('Hello world\n')
    file.write('this is a new line.\n')
```

```python
# writing a file(without overwriting)

with open('example.txt', 'a') as file:
    file.write('Appended this line\n')
```

```python
# writing a list of lines to a file

lines = ['first line\n', 'second line\n', 'third line\n']
with open('example.txt', 'a') as file:
    file.writelines(lines)
```

#### Binary Files

```python
# writing to a binary file
data = b'\x00\x01\x02\x04\n'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
# reading a binary file
with open('example.bin', 'rb') as file:
    content = file.read()
    print(content)
```

read content from source text file and writing it to destination text file

```python
# copy content from source
with open('source.txt', 'r') as source_file:
    content = source_file.read()

# paste content to destination
with open('destination.txt', 'w') as dest_file:
    dest_file.write(content)
```

Read a text file and count the number of lines, words and characters

```python
# count 
def return_count(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        words = sum(len(line.split()) for line in lines)
        char = sum(len(line) for line in lines)
    return line_count, words, char

line, word, ch = return_count('example.txt')
print(f'No.of lines:{line}, No.of Words:{word}, No.of Characters:{ch}')
    
```

```python
# writing and reading a file
with open('eg.txt', 'w+') as file:
    file.write('Hello world\n')
    file.write('This is a new line\n')

    # move the cursor to beginning
    file.seek(0)

    # read the content of the file
    content = file.read()
    print(content)
```
