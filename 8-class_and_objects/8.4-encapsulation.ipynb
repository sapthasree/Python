##### Encapsulation and Abstraction

These are the two fundamental principles of OOPs that help in designing robust, maintainable, and reusable code. Encapsulation involves bundling data and methods that operate on the data within a single unit, while abstraction involves hiding complex implementation details and exposing only the necessary features.

#### Encapsulation
It is the concept of wrapping data(variables) and methods(functions) together as a single unit. It restrict direct access to some of the object's components, which is a means of preventing accidental interference and misuse of the data

```python
# public, protected and private variables or access modifiers
```

```python
class Person:
    def __init__(self, name, age):
        # public variables
        self.name = name
        self.age = age

def print_age(person):
    return person.age

person = Person("Saptha", 21)
print(print_age(person))
```

```python
class Person:
    def __init__(self, name, age):
        # private variables (Cannot access from outside the class, even if there's a derived class)
        self.__name = name
        self.__age = age

def print_age(person):
    return person.__age

person = Person("Saptha", 21)
print(print_age(person))
```

```python
class Person:
    def __init__(self, name, age):
        # protected variables (Can be accessible only from derived class)
        self._name = name
        self._age = age

class Employee(Person):
    def __init__(self, name, age):
        super().__init__(name, age)

emp = Employee("Sapthasree", 21)
print(emp._name)
```

#### Encapsulation with getter and setter

```python
class Person:
    def __init__(self, name, age):
        self.__name = name # private access modifier
        self.__age = age # private access modifier

    # getter method for name
    def get_name(self):
        return self.__name

    # setter method for name
    def set_name(self, name):
        self.__name = name

    # getter method for age
    def get_age(self):
        return self.__age
    
    # setter method for age
    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age cannot be negative value")
```

```python
person = Person("Sapthasree", 21)

# access and modify private variables using getters and setters
print(person.get_name())
print(person.get_age())

person.set_age(22)
print(person.get_age())

person.set_age(-10)
```
