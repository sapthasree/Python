##### Magic Methods

Magic methods, also known as dunder methods (double underscore methods), are special methods that start and end with double underscores. These methods enable us to define the behavior of objects for built-in operations, such as arithmetic operations, comparisons and more.

#### Magic methods
Predefined methods in python that we can override to change the behavior of the object. Some common magic methods are:

```python
'''
__init__: initializes a new instance of a class 
__str__: returns a string representation of an object
__repr__: returns an official string representation of an object
__len__: returns the length of an object
__getitem__: gets an item from the container
__setitem__: sets an item in the container
'''
```

```python
# __init__
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

person = Person("Saptha", 21)
print(person)
```

```python
# __str__
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f'{self.name} is {self.age} years old'

person = Person("Saptha", 21)
print(person)
```

```python
# __repr__
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f'Person(name={self.name}, age={self.age})'
    
person = Person("Saptha", 21)
print(person)
```
