##### Exception handling

In python, exception handling allows to handle errors gracefully and take corrective actions without stopping the execution of the program. 

##### Exceptions

Exceptions are events that disrupt the normal flow of a program. They occur when the error is encountered during program execution. Common exceptions are:
- ZeroDivisionError: divide by zero
- FileNotFoundError: file is not present
- ValueError: invalid value
- TypeError: invalid type

```python
# NameError 
a = b
```

```python
# Exception try, except block
try:
    a = b
except:
    print('Hey, variable "b" is not assigned')
```

```python
try:
    a = b
except NameError as ex:
    print(ex)
```

```python
# ZeroDivisionError
res = 1 / 0
```

```python
try:
    res = 1 / 0
except ZeroDivisionError as ex:
    print(ex)
    print('Hey, Please enter denominator greater than "0"')
```

```python
try:
    res = 1 / 2
    a = b
except ZeroDivisionError as ex:
    print(ex)
    print('Hey, Please enter denominator greater than "0"')
```

Base Class / super class of Exception - Exception

```python
try:
    res = 1 / 2
    a = b
except ZeroDivisionError as ex:
    print(ex)
    print('Hey, Please enter denominator greater than "0"')
except Exception as ex1:
    print(ex1)
    print('Main Exception got caught here')
```

```python
# ValueError and ZeroDivisionError
try:
    num = int(input("Enter a Number: "))
    res = 10 / num
except ValueError as ex1:
    print('hey, Give a number please')
except ZeroDivisionError as ex2:
    print('hey, Please enter value greater than "0"')
```

```python
# try, except, else block
try:
    num = int(input("Enter a Number: "))
    res = 10 / num
except ValueError as ex1:
    print('hey, Give a number please')
except ZeroDivisionError as ex2:
    print('hey, Please enter value greater than "0"')
# else gets executed only if we don't have any exception. means after successful working of try block.
else:
    print(res)
```

```python
# try, except, else and finally
try:
    num = int(input("Enter a Number: "))
    res = 10 / num
except ValueError as ex1:
    print('hey, Give a number please')
except ZeroDivisionError as ex2:
    print('hey, Please enter value greater than "0"')
# else gets executed only if we don't have any exception. 
else:
    print(res)
# finally gets executed always.
finally:
    print('Execution completed.')
```

```python
# file handling and exception handling
try:
    file = open('example.txt', 'r')
    content = file.read()
    print(content)
    a = b
except FileNotFoundError as ex:
    print('File is not present')
except Exception as ex1:
    print(ex1)
finally:
    if 'file' in locals() or not file.closed():
        file.close()
        print('file closed')
```
