##### filter function

creates an iterator from elements of an iterable for which a function returns True. It is used to filter out items from a list(or any iterable) based on a condition.

```python
def even(num):
    return num % 2 == 0

print(even(4))
```

```python
lst = [1,2,5,3,6,2,1,4,1,6,3,2,8]

list(filter(even,lst))
```

```python
# filter with lambda function
nums = [2,4,5,1,7,8,5,7,10]
greater_than_five = list(filter(lambda x: x>5, nums))
print(greater_than_five)
```

```python
# filter with a lambda function and multiple conditions
nums = [2,4,5,1,7,8,5,7,10]
even_greater_than_five = list(filter(lambda x: x>5 and x%2==0, nums))
print(even_greater_than_five)
```

```python
# filter to check if the age is greater than 25 in dictionaries
people = [
    {'name':'Saptha', 'age':21},
    {'name':'Sree', 'age':20},
    {'name':'Helene', 'age':30},
    {'name':'Sapthasree', 'age':22}
]

def greater_than_25(people):
    return people['age'] > 25

list(filter(greater_than_25, people))
```

filter function is a powerful tool for creating iterators that filter items out of an iterable based on a function. It is commonly used for data cleaning, filtering objects, and removing unwanted elements from the list.
