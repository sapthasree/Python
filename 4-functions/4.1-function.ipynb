##### Function
function is a block of code that performs a specific task. Function helps in organizing the code, reusing code and improving readability.

```python
# syntax
def function_name(parameters):
    return expn
```

```python
def even_or_odd(num):
    if num % 2 == 0:
        return 'even'
    else:
        return 'odd'

print(even_or_odd(8))
```

```python
def add(a, b):
    return a+b

print(add(4, 3))
```

Variable length arguments

```python
# positional arguments
def print_numbers(*numbers):
    for num in numbers:
        print(num)

print_numbers(1,2,3,4,5,6,2,3,4)
```

```python
# keywords arguments
def print_details(**kwargs):
    for key, val in kwargs.items():
        print(f'{key}:{val}')

print_details(name="Saptha", age="20", country="India")
```

```python
# combination
def print_details_both(*args, **kwargs):
    print("Positional Arguments")
    for arg in args:
        print(arg)
    print()
    print("Keyword Arguments")
    for key, val in kwargs.items():
        print(f'{key}:{val}')

print_details_both(1,2,3,1,5,4,"Hello",name="Sree", age="20", country="India", place="Kerala" )
```
