##### map()

applies a given function to all items in an input list(or any other iterable) and returns a map object. This is used to transform the data in a list comprehensively

```python
def square(num):
    return num ** num

square(3)
```

```python
numbers = [1,2,3,4,5,6,7]

list(map(square, numbers))
```

```python
# map multiple iterables
nums1 = [1,2,3]
nums2 = [4,5,6]

list(map(lambda x,y:x+y, nums1, nums2))
```

```python
# convert a list of strings to integers
str_nums = ['1', '2', '7', '6', '5', '4']
int_nums = list(map(int, str_nums))

print(int_nums)
```

```python
words = ['apple', 'banana', 'orange', 'grape']
upp = list(map(str.upper, words))

print(upp)
```

```python
def get_name(person):
    return person['name']

person = [
    {'name':'saptha', 'age':21},
    {'name':'sree', 'age':20}
]

list(map(get_name, person))
```

map() is a powerful tool for applying transformations to iterable data structures. It can be used with regular functions, lambda functions, and even multiple iterables, providing a versatile approach to data processing in python. By understanding and utilizing map(), you can write more efficient and readable code.
