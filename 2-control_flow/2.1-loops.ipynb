##### Loops (for, while)

for loop

```python
for i in range(5):
    print(i)
```

```python
for i in range(1, 6):
    print(i)
```

```python
for i in range(1, 10, 2):
    print(i)
```

```python
for i in range(10, 1, -2):
    print(i)
```

```python
string = 'Hey Dear'

for st in string:
    print(st)
```

while loop

```python
count = 0
while count < 5:
    print(count)
    count += 1
```

Loop Control Statements

```python
# break

for i in range(5):
    if i == 3:
        break
    print(i)
```

```python
# continue

for i in range(5):
    if i == 3:
        continue
    print(i)
```

Nested loops

```python
for i in range(5):
    for j in range(1,3):
        print(i, j)
```

```python
## sum of first n natural numbers using a while and for loop
```

```python
# while loop

num = 5
total = 0
count = 1

while count <= num:
    total += count
    count += 1

print(f"Sum of first {num} no.s = {total}")
```

```python
# for loop

num = 5
total = 0

for i in range(num+1):
    total += i

print(f"Sum of first {num} no.s = {total}")
```

```python
## prime no.s between 1 and 100

for num in range(1, 101):
    if num > 1:
        for i in range(2, int(num ** 0.5)+1):
            if num % i == 0:
                break
        else:
            print(num, end=',')
```
