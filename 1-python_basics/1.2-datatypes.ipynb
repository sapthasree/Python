#### Data Types
classification of data which tell the compiler or interpreter how the programmer intends to use the data

```python
## integer

a = 10
print(a, type(a))
```

```python
## floating point

b = 5.1
print(b, type(b))
```

```python
## string

s = 'hello'
print(s, type(s))
```

```python
boolean = True
print(boolean, type(boolean))
```
