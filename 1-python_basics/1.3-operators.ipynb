#### Operators

```python
## Arithmetic Operations

a = 10
b = 5
```

```python
## 1. addition

print("Sum: ", a+b)
```

```python
## 2. Subtraction

print("Subtraction: ", a-b)
```

```python
## 3. Multiplication

print("Multiplication: ", a*b)
```

```python
## 4. Division

print("Floor Division: ", a//b)
print("Integer Division: ", a/b)
print("Modular Division: ", a%b)
```

```python
## 5. Exponential

print("Exponential: ", a**b)
```

```python
## Comparison Operators

a = 10
b = 10
```

```python
## == operator
print(a==b)
```

```python
## != operator
print(a!=b)
```

```python
## > operator
print(a>b)
print(b>a)
```

```python
## >= operator
print(a>=b)
print(b>=a)
```

```python
## < operator
print(a<b)
print(b<a)
```

```python
## <= operator
print(a<=b)
print(b<=a)
```

```python
## Logical Operator

x = True
y = False
```

```python
## AND operator

print(x and y)
```

```python
## OR operator

print(x or y)
```

```python
## NOT operator

print(not x)
print(not y)
```
