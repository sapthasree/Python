##### Variables
used to store the data that can be referenced and manipulated in a program.

```python
## declaring and assigning variables
age = 21
height = 5.0
name = "Sapth"
is_student = True

print("age:", age)
print("height:", height)
print("name:", name)
print("is_student:", is_student)
```

```python
## naming conventions

## 1. varname should be descriptive
## 2. it must start with a letter or an underscore
## 3. can contain letters, numbers and underscore in between
## 4. varnames are case sensitive

## valid varName
first_name = "Sapth"

## invalid varName
## 2Age = 20
## @name = 'Krish'
## fst-name = 'Hello'
```

```python
## Type checking and conversions
print(height, type(height))

height_str = str(height)
print(height_str, type(height_str))
```

```python
## ValueError
print(name, type(name))

name_int = int(name)
print(name_int, type(name_int))
```

```python
## dynamic typing
## python allows the type of a variable to change as it is executing
var = 10
print(var, type(var))

var = 'hello'
print(var, type(var))

var = 3.14
print(var, type(var))
```

```python
## input

age = input("What is the age?")
print(age, type(age))
```

### Simple Calculator

```python
num1 = float(input("Enter number 1: "))
num2 = float(input("Enter number 2: "))

sum = num1 + num2
diff = num1 - num2
pdt = num1 * num2
quotient = num1 / num2

print("Num1: ", num1)
print("Num2: ", num2)

print("Sum: ", sum)
print("Difference: ", diff)
print("Product: ", pdt)
print("Quotient: ", quotient)
```
